function isPositiveAndEven(num) {
  let isPositive = num > 0;        
  let isOdd = num % 2 === 1;
  return isPositive && isOdd;           
}

// Test it with an input of 8
let testNum = 8;

console.log("Testing number:", testNum);
console.log("Is it positive?", testNum > 0);
console.log("Is it Odd?", testNum % 2 === 1);
console.log("Final result (positive AND odd):", isPositiveAndEven(testNum));
