<h2>Night at The Museum!</h2>

<p>
I attended Night at The Museum where I collaborated with peers from my CSSE Period 1 Class, and we presented our Murder Mystery game.
</p>

<h3>My Experience</h3>

<p>
It was really interesting to look at the projects of other students who are also interested in computer science and how projects in higher grades look like. I was also able to present the work I worked on the past few weeks to parents which was fun.
</p>

<h3>Below are some of the computer science projects that I found interesting.</h3>

<div class="natm-grid">

  <div class="natm-card">
    <img src="{{site.baseurl}}/images/project1.png">
    <p>
    This group made an interactive poll asking which AI they find most helpful. From this, I learned that the coders used a special API and database to keep count of the votes.
    </p>
    <em>Project 1</em>
  </div>

  <div class="natm-card">
    <img src="{{site.baseurl}}/images/project2.png">
    <p>
    This group made a trip planner and a new way to plan a vacation. I learned that they also used API to list landmarks, shopping, and weather probabilities.
    </p>
    <em>Project 2</em>
  </div>

  <div class="natm-card">
    <img src="{{site.baseurl}}/images/project3.png">
    <p>
    This group made a snakes and ladders game that players use to study for the APCSP exam. The game starts by completing 6 minigames to get a grasp of all the concepts. Then they move to the game board where they answer mini quizzes. At the end they play a boss battle to finish the game. They used websockets, and local storage to keep track of scores and the functionality of the game.
    </p>
    <em>Project 3</em>
  </div>

</div>
